```python
# Nöron'un değeri = ( x1*w1 + x2*w2 + b ) ve sıkıştırma işlemi
# x1 ve x2 inputlarımız. w1 ve w2 ağırlıklarımız. b bias değerimiz.
# ağırlıklar nöral ağlarda nöronlar arası bağlantıların sonucu ne kadar
# etkilediği gösteriyor.

import math #tanh için
class Value:
  def __init__(self, data):
   self.data = data

  def __add__(self, other):
    return Value (self.data + other.data)

  def __mul__(self, other):
    return Value (self.data * other.data)

x1 = Value (2.0)
x2 = Value (4.0)
w1 = Value (1.0)
w2 = Value (1.0)
b = Value  (1.0)
n = x1*w1 + x2*w2 + b
n.data
math.tanh(n.data)

```
